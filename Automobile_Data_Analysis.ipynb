```python
auto.info()
```

## PREPROCESSING

```python
auto.replace('?', np.nan, inplace=True)
auto.info()
```

```python
auto.isnull().sum()
```

```python
auto.nunique()
```

```python
auto['normalized-losses']=auto['normalized-losses'].apply(float)
auto['price']=auto['price'].apply(float)
auto['bore']=auto['bore'].apply(float)
auto['stroke']=auto['stroke'].apply(float)
auto['horsepower']=auto['horsepower'].apply(float)
auto['peak-rpm']=auto['peak-rpm'].apply(float)
```

```python
auto.info()
```

```python
auto.duplicated().sum()
```

```python
avg_normalized_losses =auto['normalized-losses'].mean()
auto['normalized-losses'].replace(np.nan , avg_normalized_losses, inplace= True)

avg_price =auto['price'].mean()
auto['price'].replace(np.nan , avg_price, inplace= True)
```

```python
auto.isnull().sum()
```

```python
fre_door = auto['num-of-doors'].dropna().mode()[0]
auto['num-of-doors'].replace(np.nan , fre_door, inplace= True)

fre_bore = auto['bore'].dropna().mode()[0]
auto['bore'].replace(np.nan , fre_bore, inplace= True)

fre_stroke = auto['stroke'].dropna().mode()[0]
auto['stroke'].replace(np.nan , fre_stroke, inplace= True)

fre_horsepower = auto['horsepower'].dropna().mode()[0]
auto['horsepower'].replace(np.nan , fre_horsepower, inplace= True)

fre_peak_rpm = auto['peak-rpm'].dropna().mode()[0]
auto['peak-rpm'].replace(np.nan ,fre_peak_rpm, inplace= True)
```

```python
auto.isnull().sum()
```
